```python
# Using matplotlib t,o create a count plot
#UD
quality_counts = df['quality'].value_counts()
plt.figure(figsize=(8,6))
plt.bar(quality_counts.index,quality_counts,color='red')
plt.title('Count Plot of Quality')
plt.xlabel('Qualtity')
plt.ylabel('Count')
plt.show
# print the values
# print(quality_counts)
```

```python
# Using matplotlib t,o create a count plot
#UD
# Categorical scatter plot for outliers
plt.figure(figsize=(10,8))
#Using seaborn to create a swarm plot
sns.swarmplot(x="quality",y="alcohol",data=df,palette="viridis")
plt.title('Swarm Plot for Quality and Alcohol')
plt.xlabel('Qualtity')
plt.ylabel('Alcohol')
plt.show
```

```python
#BD
# estblish alcohol and qualities
# Categorical scatter plot for outliers
# df['quality'] = df['quality'].astype(str)
plt.figure(figsize=(10,8))
#Using seaborn to create a swarm plot
sns.violinplot(x="quality",y="alcohol",data=df,palette="viridis",alpha=0.9)
plt.title('Violin Plot for Quality and Alcohol')
plt.xlabel('Qualtity')
plt.ylabel('Alcohol')
plt.show
```

```python
# alternative version of violin and boxplot
sns.boxplot(x='quality',y='alcohol',data=df)
plt.show()
```

```python
# Categorical box plot
sns.boxenplot(x='quality',y='alcohol',data=df)
plt.show()
```

```python
# MD
plt.figure(figsize=(15,10))

sns.heatmap(df.corr(), annot=True, fmt = '.4f', cmap='Pastel1', linewidths=10)

plt.title("Corealtion Heatmap")
plt.show()
```
